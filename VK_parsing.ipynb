```python
data2 = pd.DataFrame()
for group_id in group_id_list:
    print("group:",group_id)
    for age in range(18,25):
        res_users = requests.get('https://api.vk.com/method/users.search',
                             params={
                                 'access_token' : TOKEN,
                                 'v' : API_VERSION,
                                 'offset' : 0,
                                 'count' : 1000,
                                 'fields' : 'bdate, sex, crop_photo',
                                 'university' : 297,
                                 'age_from' : age,
                                 'age_to' : age,
                                 'has_photo' : 1,
                                 'group_id' : group_id
                             })
        pd_users = pd.DataFrame(res_users.json()['response']['items'])
        data2 = pd.concat([data2, pd_users])
        print("Data len:",len(data2))
data2.reset_index(drop=True, inplace = True)
```
